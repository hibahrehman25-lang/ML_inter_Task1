```python
con.sql("""
SELECT
    content_hash_id,
    COUNT(DISTINCT report_date) AS days_present
FROM 'hf://datasets/FlyRank/internship-warehouse/fact_content_daily_performance/month=2026-03/data_0.parquet'
GROUP BY content_hash_id
ORDER BY days_present DESC
LIMIT 10
""").df()
```

```python
con.sql("""
SELECT
    COUNT(*) AS total_rows,
    COUNT(gsc_impressions) AS impressions_not_null,
    COUNT(gsc_clicks) AS clicks_not_null,
    COUNT(gsc_avg_position) AS position_not_null,
    COUNT(scroll_events) AS scroll_not_null
FROM 'hf://datasets/FlyRank/internship-warehouse/fact_content_daily_performance/month=2026-03/data_0.parquet'
WHERE gsc_data_available IS TRUE
""").df()
```

```python
con.sql("""
SELECT
    gsc_impressions,
    gsc_clicks,
    gsc_avg_position
FROM 'hf://datasets/FlyRank/internship-warehouse/fact_content_daily_performance/month=2026-03/data_0.parquet'
WHERE gsc_data_available IS TRUE
LIMIT 10
""").df()
```

```python
con.sql("""
SELECT
    MIN(report_date) AS first_date,
    MAX(report_date) AS last_date
FROM 'hf://datasets/FlyRank/internship-warehouse/fact_content_daily_performance/month=2026-03/data_0.parquet'
""").df()
```

```python
con.sql("""
SELECT
    content_hash_id,
    gsc_impressions,
    gsc_clicks,
    ROUND(
        CASE
            WHEN gsc_impressions > 0
            THEN (gsc_clicks * 100.0 / gsc_impressions)
            ELSE NULL
        END,2
    ) AS ctr_percent
FROM 'hf://datasets/FlyRank/internship-warehouse/fact_content_daily_performance/month=2026-03/data_0.parquet'
WHERE
    gsc_data_available IS TRUE
    AND gsc_impressions >= 100
ORDER BY ctr_percent ASC
LIMIT 10;
""").df()
```

```python
con.sql("""
SELECT
    AVG(gsc_impressions) AS avg_impressions,
    AVG(gsc_clicks) AS avg_clicks,
    AVG(gsc_avg_position) AS avg_position
FROM 'hf://datasets/FlyRank/internship-warehouse/fact_content_daily_performance/month=2026-03/data_0.parquet'
WHERE gsc_data_available IS TRUE
""").df()
```

## 1. Unit of Analysis + Time Window

### Unit of Analysis
One row represents the daily search performance of one content page (`content_hash_id`) for one client (`client_hash_id`) on one reporting date (`report_date`).

### Tables Used
- `fact_content_daily_performance`
- `dim_content` (used only if additional content metadata is needed)

### Time Window
This analysis uses the **March 2026** partition (`month = 2026-03`), covering data from **2026-03-01** to **2026-03-31**.

### Prediction / Ranking Objective
My selected lane is **CTR / Engagement Opportunity Scoring**. The goal is to rank content pages that receive high Google Search impressions but relatively few clicks so they can be prioritized for CTR optimization.

### Deliberately Excluded
I excluded AI referral metrics because they are outside the scope of my CTR opportunity task. I also excluded most GA4 metrics because they are available for only a subset of rows, making them unsuitable as primary features for this notebook.

## Feature Notes

### gsc_avg_position

### scroll_events

### client_has_gsc
Known before analysis and indicates whether Google Search Console data exists for the client.

### gsc_data_available
Used to ensure only valid GSC observations are included in the analysis.

## Leakage Check

As an experiment, I considered using `gsc_clicks` as an input feature.

Because the objective is to identify pages with high impressions but relatively low clicks, including `gsc_clicks` would provide information that is directly related to the ranking objective and could artificially improve model performance.

To avoid leakage, I excluded `gsc_clicks` from the final feature set and retained only features that would be available independently at the decision moment.
