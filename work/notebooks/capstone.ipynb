```python
# ============================================================
# WEEK 8 — CAPSTONE
# Fresh Colab setup
# ============================================================

import os
import sys
import subprocess
from pathlib import Path

# ------------------------------------------------------------
# 1. Install only the packages required by this notebook
# ------------------------------------------------------------
subprocess.check_call([
    sys.executable, "-m", "pip", "install", "-q",
    "duckdb",
    "huggingface_hub",
    "scikit-learn",
    "matplotlib",
    "pandas",
    "numpy"
])

# ------------------------------------------------------------
# 2. Locate the repository
#    This works whether Colab opened the repo directly
#    or the repo already exists under /content.
# ------------------------------------------------------------
REPO_ROOT = None

search_roots = [
    Path("/content"),
    Path.cwd()
]

for search_root in search_roots:
    if search_root.exists():
        matches = list(search_root.rglob("capstone.ipynb"))
        for match in matches:
            candidate = match.parent.parent.parent
            if (
                (candidate / "work").exists()
                and (candidate / "submission").exists()
            ):
                REPO_ROOT = candidate
                break
    if REPO_ROOT is not None:
        break

if REPO_ROOT is None:
    raise FileNotFoundError(
        "Repository root could not be located. "
        "Open capstone.ipynb from the GitHub repository in Colab first."
    )

os.chdir(REPO_ROOT)

print("Repository root:", REPO_ROOT)
print("Current working directory:", Path.cwd())

# ------------------------------------------------------------
# 3. Create the required artifact/output directories
# ------------------------------------------------------------
OUTPUT_DIR = REPO_ROOT / "work" / "outputs"
PAPER_DIR = REPO_ROOT / "docs"

OUTPUT_DIR.mkdir(parents=True, exist_ok=True)
PAPER_DIR.mkdir(parents=True, exist_ok=True)

print("Output directory:", OUTPUT_DIR)
print("Paper directory:", PAPER_DIR)

# ------------------------------------------------------------
# 4. Imports
# ------------------------------------------------------------
import duckdb
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from google.colab import userdata

from sklearn.model_selection import (
    train_test_split,
    GroupShuffleSplit
)

from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import HistGradientBoostingClassifier

from sklearn.metrics import (
    precision_score,
    recall_score,
    f1_score,
    accuracy_score,
    average_precision_score,
    precision_recall_curve
)

# ------------------------------------------------------------
# 5. Reproducibility
# ------------------------------------------------------------
RANDOM_STATE = 42
np.random.seed(RANDOM_STATE)

# ------------------------------------------------------------
# 6. Hugging Face authentication
#    Token is read from Colab Secrets.
#    Never print or save the token.
# ------------------------------------------------------------
HF_TOKEN = userdata.get("HF_TOKEN")

if not HF_TOKEN:
    raise ValueError(
        "HF_TOKEN was not found in Colab Secrets. "
        "Add HF_TOKEN to Colab Secrets and run this cell again."
    )

# ------------------------------------------------------------
# 7. DuckDB connection
# ------------------------------------------------------------
con = duckdb.connect()

con.execute(f"""
CREATE OR REPLACE SECRET hf_secret (
    TYPE HUGGINGFACE,
    TOKEN '{HF_TOKEN}'
);
""")

print("DuckDB connection: ready")
print("Hugging Face authentication: ready")

# ------------------------------------------------------------
# 8. Project-level constants
# ------------------------------------------------------------
MODEL_MONTH = "2026-03"
FINAL_SEALED_MONTH = "2026-06"

print("Modeling window:", MODEL_MONTH)
print("Final sealed month:", FINAL_SEALED_MONTH)

print("\nSetup completed successfully.")
```
