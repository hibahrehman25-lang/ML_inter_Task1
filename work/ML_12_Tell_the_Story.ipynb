# ML-12 — Tell the Story

## 5-Minute Demo Outline

### 1. Question — 1 minute

FlyRank content teams cannot manually review every content page with the same level of attention. The practical question is which existing pages should be reviewed first when search visibility is present but performance may be declining.

This work treats the problem as **content-review prioritization**, not as a claim that a model can predict whether a future content change will improve performance.

### 2. Method — 1 minute

The workflow treats the task as an opportunity-ranking problem. Decision Tree, Logistic Regression, and Random Forest were evaluated against a fixed-rule baseline.

The primary selection metric was **Precision@50**, because the practical output is a ranked queue in which reviewers inspect the highest-priority pages first. Validation used a **client-holdout split**, supported by leakage checks and signal audits.

### 3. One Chart — 1 minute

**Chart to show:** Model comparison / Precision@50.

The key comparison is:

* Random Forest: **0.740 Precision@50**
* Fixed-rule baseline: **0.240 Precision@50**

The chart communicates the main result: the learned ranking model placed more labeled declining opportunities near the top of the review queue than the fixed-rule baseline on the same validation setup.

### 4. One Honest Result — 1 minute

The Random Forest placed **37 of the top 50** ranked items among the labeled declining opportunities, compared with **12 of the top 50** under the baseline rules.

This is a validation result showing stronger ranking performance. It does **not** prove that refreshing those pages will increase traffic, clicks, CTR, rankings, or engagement.

### 5. One Recommendation — 1 minute

Use the ranked queue to help content reviewers decide where to look first. Start with high-confidence opportunities, inspect the reason codes and page context, and make the final decision manually.

The model should **not** automatically publish, rewrite, delete, or replace content.
